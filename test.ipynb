```python
def is_palindrome(s):
    new_s = "".join(s.lower().split())
    half = len(s)//2
    count = 1
    for index, char in enumerate(new_s[:half]):
        if char != new_s[-(index+1)]:
            count -= 1
            return False
    if count == 1: 
        return True
        
```

```python
print(is_palindrome('Eleven animals I slam in a net'))
```

```python
from collections import Counter
def is_anagram(s1, s2):
    new_s1 = "".join(s1.lower().split())
    new_s2 = "".join(s2.lower().split())

    char_counts_s1 = Counter(new_s1)
    char_counts_s2 = Counter(new_s2)

    if char_counts_s1 == char_counts_s2:
        return True 
    else:
        return False


```

```python
s1 = 'modern'
s2 = 'norman'

is_anagram(s1, s2)
```

```python
def define_tile_of_point(x,y):

    if x in range(3) and y in range(3):
        return 'Tile 1'
    if x in range(3) and y in range(3,6):
        return 'Tile 2'
    if x in range(3) and y in range(6,9):
        return 'Tile 3'
    
    if x in range(3,6) and y in range(3):
        return 'Tile 4'    
    if x in range(3,6) and y in range(3,6):
        return 'Tile 5'    
    if x in range(3,6) and y in range(6,9):
        return 'Tile 6'
    
    if x in range(6,9) and y in range(3):
        return 'Tile 7'
    if x in range(6,9) and y in range(3,6):
        return 'Tile 8'
    if x in range(6,9) and y in range(6,9):
        return 'Tile 9'
    
    
    
```

```python
import numpy as np
line = [list(map(int, list(s))) for s in input().split()]
arr = np.reshape(line, (9,9))
```

```python
def check_sudoku(arr):
    for i in range(1, 10):
        indices = np.argwhere(arr == i)
        list_tile = []
        x_axis = []
        y_axis = []

        for idx in indices:
            x, y = idx
            result = define_tile_of_point(x, y)

            if x in x_axis: 
                return 'Not a valid Sudoku'
            if y in y_axis: 
                return 'Not a valid Sudoku'
            if result in list_tile:
                return 'Not a valid Sudoku'
            
            list_tile.append(result)
            x_axis.append(x)
            y_axis.append(y)

    return 'A valid Sudoku'



    
```

```python
check_sudoku(arr)
    
```
