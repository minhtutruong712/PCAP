```python
def is_palindrome(s):
    new_s = "".join(s.lower().split())
    half = len(s)//2
    count = 1
    for index, char in enumerate(new_s[:half]):
        if char != new_s[-(index+1)]:
            count -= 1
            return False
    if count == 1: 
        return True
        
```

```python
print(is_palindrome('Eleven animals I slam in a net'))
```

```python
from collections import Counter
def is_anagram(s1, s2):
    new_s1 = "".join(s1.lower().split())
    new_s2 = "".join(s2.lower().split())

    char_counts_s1 = Counter(new_s1)
    char_counts_s2 = Counter(new_s2)

    if char_counts_s1 == char_counts_s2:
        return True 
    else:
        return False


```

```python
s1 = 'modern'
s2 = 'norman'

is_anagram(s1, s2)
```

```python
def check_sudoku(arr):
    for i in range(1, 10):
        indices = np.argwhere(arr == i)
        list_tile = []
        x_axis = []
        y_axis = []

        for idx in indices:
            x, y = idx
            result = define_tile_of_point(x, y)

            if x in x_axis: 
                return 'Not a valid Sudoku'
            if y in y_axis: 
                return 'Not a valid Sudoku'
            if result in list_tile:
                return 'Not a valid Sudoku'
            
            list_tile.append(result)
            x_axis.append(x)
            y_axis.append(y)

    return 'A valid Sudoku'



    
```

```python
def fibonacci(n):
    p = pp = 1
    for i in range(n):
        if i in [0, 1]:
            yield 1
        else:
            n = p + pp
            pp, p = p, n
            yield n

fibs = list(fibonacci(10))
print(fibs)
```
